### Multithreading
## When to use Multi Threading
## I/O-bound tasks: Tasks that spend more time waiting for I/O operations (e.g., file operations, network requests).
###  Concurrent execution: When you want to improve the throughput of your application by performing multiple operations concurrently.

```python
import threading
import time
```

# basic operation without using the mutiple threading

```python
def print_numbers():
    for number in range(0,5):
        time.sleep(2)
        print(number)

def print_letters():
    for letter in 'abcde':
        time.sleep(2)
        print(letter)

t = time.time()

print_letters()
print_numbers()
finished_time= time.time()-t

print(finished_time)
```

# Opeartion using the Multi-threading

```python
def print_numbers():
    for number in range(0,5):
        time.sleep(2)
        print(f"{number}")

def print_letters():
    for letter in 'abcde':
        time.sleep(2)
        print(f"{letter}")

## create Two threads
t1 = threading.Thread(target = print_numbers)
t2 = threading.Thread(target = print_letters)


t = time.time()
## start the thread
t1.start()
t2.start()

## wait for the thread to complete
t1.join()
t2.join()

finished_time= time.time()-t

print(finished_time)
```
