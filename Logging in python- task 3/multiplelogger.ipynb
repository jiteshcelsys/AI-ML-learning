# Logging with Mutiple Loggers

You can create mutiple loggers for different parts of your application

```python
import logging
## create a logger for module1
logger1 = logging.getLogger('Module1')
logger1.setLevel(logging.DEBUG)

## create a logger for module2
logger2 = logging.getLogger('Module2')
logger2.setLevel(logging.WARNING)


## Configurinn logging settings
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s- %(name)s- %(levelname)s - %(message)s',
    datefmt='%Y- %m-%d %H: %M: %S'
)

```

```python
logger1.debug('This is debug message for module 1')
logger2.warning('This is an error warning for module 2')
logger2.error('This is an error message for module 2')
```
